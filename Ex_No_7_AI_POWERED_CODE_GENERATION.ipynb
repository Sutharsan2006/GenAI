```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

# --------------------------------------------------
# Load CodeGen Model
# --------------------------------------------------

model_name = "Salesforce/codegen-350M-mono"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

# --------------------------------------------------
# Code Generation Function
# --------------------------------------------------

def generate_code(prompt, max_new_tokens=80):

    input_ids = tokenizer(
        prompt,
        return_tensors="pt"
    ).input_ids

    output = model.generate(
        input_ids,
        max_new_tokens=max_new_tokens,
        pad_token_id=tokenizer.eos_token_id,
        do_sample=False
    )

    return tokenizer.decode(
        output[0],
        skip_special_tokens=True
    )


# --------------------------------------------------
# 1. Code Generation from Natural Language
# --------------------------------------------------

prompt1 = """# Write a Python function to check if a number is prime
def is_prime(n):"""

print("Generated Function:")
print(generate_code(prompt1))


# --------------------------------------------------
# 2. Debugging a Faulty Code Snippet
# --------------------------------------------------

buggy_code = """# The following function should return the factorial of n,
# but has a bug. Fix it.

def factorial(n):
    result = 0
    for i in range(1, n + 1):
        result = result * i
    return result

# Corrected function:

def factorial_fixed(n):"""

print("\nDebug Suggestion:")
print(generate_code(buggy_code, max_new_tokens=60))
```
